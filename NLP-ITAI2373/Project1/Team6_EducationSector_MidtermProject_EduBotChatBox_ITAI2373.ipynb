#**Section 3: Preprocessing - Data Analysis and Knowledge Base Creation**:
Contribution: Natalia Solorzano Perez W207818526

```python
# Display dataset columns
print("Dataset Columns:")
print(df.columns)

# Basic Information about the dataset
print("\nDataset Info:")
print(df.info())

# Basic Statistics
print("\nStatistical Summary:")
print(df.describe())

# Let's check for missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Analyzing important insights from data
# (Example: Compare average scores based on test preparation course)

print("\nAverage scores based on Test Preparation Course:")
print(df.groupby('test preparation course')[['math score', 'reading score', 'writing score']].mean())

# Now, create a small knowledge base of sentences based on the dataset findings
knowledge_base = [
    "Students who completed the test preparation course scored significantly higher in math, reading, and writing compared to those who did not.",
    "On average, female students performed better in reading and writing than male students.",
    "Math scores were relatively similar between genders, but slightly higher for males.",
    "Lunch type affected scores: students with standard lunch performed better than those with free/reduced lunch.",
    "Parental level of education shows a positive influence on students' performance: the higher the parent's education, the higher the student's scores.",
    "Students from group E (ethnic group) achieved the highest average scores among all groups.",
    "Completion of a test preparation course notably improved writing scores.",
    "There is a strong correlation between reading and writing scores among all students."
]

# Display the knowledge base
print("\nKnowledge Base Sentences:")
for i, fact in enumerate(knowledge_base, 1):
    print(f"{i}. {fact}")
```

#**Section 4: Building TF-IDF Model & EduBot Chatbot Engine Creation**
Contribution: Natalia Solorzano Perez W207818526

```python
# 1. Import TF-IDF Vectorizer and Cosine Similarity from sklearn
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# 2. Create the TfidfVectorizer instance
vectorizer = TfidfVectorizer()

# 3. Fit and transform the knowledge base sentences
tfidf_matrix = vectorizer.fit_transform(knowledge_base)

# 4. Define a function to find the best matching response
def get_response(user_input):
    user_tfidf = vectorizer.transform([user_input])  # Vectorize the user's input
    similarities = cosine_similarity(user_tfidf, tfidf_matrix)  # Compare to all knowledge base sentences
    index = similarities.argmax()  # Find the index of the most similar sentence
    confidence = similarities.max()  # Get the similarity score

    if confidence < 0.2:
        return "I'm sorry, I don't have enough information about that topic yet."
    else:
        return knowledge_base[index]

# 5. Build a simple chatbot loop
def chat():
    print("Hello! I'm EduBot, your educational assistant. Ask me anything about student performance!")
    print("Type 'exit' to end the conversation.\n")

    while True:
        user_input = input("You: ")
        if user_input.lower() == 'exit':
            print("EduBot: Goodbye! Keep learning and exploring!")
            break
        response = get_response(user_input)
        print("EduBot:", response)

# 6. Start the chatbot
chat()
```
